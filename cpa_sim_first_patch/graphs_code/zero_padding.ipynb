```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Plot-only notebook: the numbers are produced by `python postprocess.py` (-> ./data/*.csv)
os.makedirs('../plots', exist_ok=True)
```

```python
df = pd.read_csv('data/zero_padding.csv')
```

```python
# correlation between the HW hypotheses of -w and w (w = 0.111511) on fc2 inputs, vs. zeros padded
cors = df['correlation'].tolist()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fontSize = 32
# your data
x_vals = np.array([0, 1, 10, 50, 100, 150, 200, 1000, 10000, 100000, 1000000, 10_000_000, 100_000_000])

fig, ax = plt.subplots(figsize=(14, 10))

ax.plot(x_vals, cors, marker='o', linestyle='-', markersize=16)

# use symlog so zero can appear
ax.set_xscale('symlog', linthresh=1)   # linthresh controls the linear region around zero

# set all ticks including zero
ax.set_xticks([0, 1, 10, 100, 1000, 10000, 100000, 1000000, 10_000_000, 100_000_000])


ax.set_xlabel('Zero Padding Added', fontsize = fontSize)
ax.set_ylabel('Correlation', fontsize = fontSize)
ax.grid(True, which='both', linestyle='--', linewidth=2)
ax.set_xticklabels(['0', '1', '10', '100', '1k', '10k', '100k', '1M', '10M', '100M'], fontsize=fontSize, rotation=45, ha='center')
ax.legend()
plt.yticks(fontsize=fontSize)
fig.savefig("../plots/zero_padding.pdf",
            format="pdf",
            bbox_inches="tight",
            metadata={"Creator": "Matplotlib"})
plt.tight_layout()
plt.show()
```
